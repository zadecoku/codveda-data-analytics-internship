```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(iris["petal_length"], kde=True)
plt.title("Shpërndarja e Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Numri i mostrave")
plt.show()
```

```python
sns.histplot(iris["sepal_length"], kde=True)
plt.title("Shpërndarja e Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Numri i mostrave")
plt.show()
```

```python
sns.histplot(iris["sepal_width"], kde=True)
plt.title("Shpërndarja e Sepal Width")
plt.xlabel("Sepal Width (cm)")
plt.ylabel("Numri i mostrave")
plt.show()
```

```python
sns.histplot(iris["petal_width"], kde=True)
plt.title("Shpërndarja e Petal Width")
plt.xlabel("Petal Width (cm)")
plt.ylabel("Numri i mostrave")
plt.show()
```

```python
sns.boxplot(data=iris, x="species", y="petal_length")
plt.title("Petal Length sipas Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.show()
```

```python
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species")
plt.title("Petal Length vs Petal Width sipas Species")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.show()
```
