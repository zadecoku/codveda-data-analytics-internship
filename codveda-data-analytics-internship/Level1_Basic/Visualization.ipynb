```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

iris = pd.read_csv(r"C:\Users\Personal\Downloads\Data Set For Task-20260922T064326Z-1-001\Data Set For Task\1) iris.csv")
```

```python
avg_petal = iris.groupby("species")["petal_length"].mean()
avg_petal
```

```python
avg_petal.plot(kind="bar", color=["#66c2a5", "#fc8d62", "#8da0cb"])
plt.title("Mesatarja e Petal Length sipas Species")
plt.xlabel("Species")
plt.ylabel("Petal Length mesatare (cm)")
plt.show()
```

```python
sorted_iris = iris.sort_values("sepal_length").reset_index(drop=True)

plt.plot(sorted_iris.index, sorted_iris["sepal_length"], label="Sepal Length")
plt.plot(sorted_iris.index, sorted_iris["petal_length"], label="Petal Length")
plt.title("Sepal Length vs Petal Length (mostrat të renditura sipas Sepal Length)")
plt.xlabel("Mostra (e renditur)")
plt.ylabel("Gjatësia (cm)")
plt.legend()
plt.savefig("line_length_trend.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
sns.scatterplot(data=iris, x="sepal_length", y="sepal_width", hue="species")
plt.title("Sepal Length vs Sepal Width sipas Species")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.legend(title="Species")
plt.savefig("scatter_sepal.png", dpi=150, bbox_inches="tight")
plt.show()
```
